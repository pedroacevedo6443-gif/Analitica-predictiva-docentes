```python
import numpy as np
import pandas as pd

# Vectores con los valores
sp500 = np.array([2912.43, 3044.31, 3100.29, 3271.12, 3500.31, 3363.00,
                  3269.96, 3621.63, 3756.07, 3714.24, 3811.15, 3943.34])

nike = np.array([87.18, 98.58, 98.05, 97.61, 111.89, 125.54,
                 120.08, 134.70, 141.47, 133.59, 134.78, 140.45])

# Vector de numeración del 1 al 12
periodo = np.arange(1, 13)

# Tabla: numeración | S&P 500 | Nike Stock Price
tabla = pd.DataFrame({
    "No.": periodo,
    "S&P 500": sp500,
    "Nike Stock Price": nike
})

tabla
```

```python
# Tabla de cálculo: x = S&P 500, y = Nike Stock Price
calc = pd.DataFrame({
    "x": tabla["S&P 500"],
    "y": tabla["Nike Stock Price"]
})
calc["x*y"] = calc["x"] * calc["y"]
calc["x²"] = calc["x"] ** 2
calc["y²"] = calc["y"] ** 2

# Sumatorias de cada columna
sumas = calc.sum()
calc_tabla = pd.concat([calc, pd.DataFrame([sumas], index=["Σ"])])
display(calc_tabla)

# Valores obtenidos con pandas
n = len(calc)
sum_x, sum_y = sumas["x"], sumas["y"]
sum_xy, sum_x2, sum_y2 = sumas["x*y"], sumas["x²"], sumas["y²"]

# Coeficiente de correlación de Pearson:
# r = (nΣxy − ΣxΣy) / √[(nΣx² − (Σx)²)(nΣy² − (Σy)²)]
numerador = n * sum_xy - sum_x * sum_y
denominador = np.sqrt((n * sum_x2 - sum_x ** 2) * (n * sum_y2 - sum_y ** 2))
r = numerador / denominador

print(f"n    = {n}")
print(f"Σx   = {sum_x:.2f}")
print(f"Σy   = {sum_y:.2f}")
print(f"Σxy  = {sum_xy:.2f}")
print(f"Σx²  = {sum_x2:.2f}")
print(f"Σy²  = {sum_y2:.2f}")
print(f"r    = {r:.4f}")
print(f"r²   = {r**2:.4f}")
print(f"Verificación con numpy: {np.corrcoef(sp500, nike)[0, 1]:.4f}")
```

```python
import matplotlib.pyplot as plt

# Regresión lineal y R²
m, b = np.polyfit(sp500, nike, 1)
r2 = np.corrcoef(sp500, nike)[0, 1] ** 2
x_line = np.linspace(sp500.min(), sp500.max(), 100)

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(10, 6))

# Puntos coloreados por periodo (1 a 12) con tamaño grande
sc = ax.scatter(sp500, nike, c=periodo, cmap="plasma", s=220,
                edgecolor="white", linewidth=1.5, zorder=3)

# Línea de tendencia con sombra de brillo
ax.plot(x_line, m * x_line + b, color="#00e5ff", linewidth=3, zorder=2,
        label=f"Tendencia: y = {m:.4f}x + {b:.2f}  (R² = {r2:.3f})")
ax.fill_between(x_line, m * x_line + b - 6, m * x_line + b + 6,
                color="#00e5ff", alpha=0.12, zorder=1)

# Numeración de cada punto
for i, (x, y) in enumerate(zip(sp500, nike), start=1):
    ax.annotate(str(i), (x, y), ha="center", va="center",
                fontsize=9, fontweight="bold", color="white", zorder=4)

cbar = plt.colorbar(sc, ax=ax, pad=0.02)
cbar.set_label("Periodo", fontsize=12)

ax.set_xlabel("S&P 500", fontsize=13, fontweight="bold")
ax.set_ylabel("Nike Stock Price (USD)", fontsize=13, fontweight="bold")
ax.set_title("Nike Stock Price vs S&P 500", fontsize=18, fontweight="bold", pad=15)
ax.grid(True, linestyle="--", alpha=0.25)
ax.legend(fontsize=11, loc="upper left", frameon=True)
for s in ("top", "right"):
    ax.spines[s].set_visible(False)

plt.tight_layout()
plt.show()
```
