```python
import numpy as np
import pandas as pd

# Vectores con los valores
sp500 = np.array([2912.43, 3044.31, 3100.29, 3271.12, 3500.31, 3363.00,
                  3269.96, 3621.63, 3756.07, 3714.24, 3811.15, 3943.34])

nike = np.array([87.18, 98.58, 98.05, 97.61, 111.89, 125.54,
                 120.08, 134.70, 141.47, 133.59, 134.78, 140.45])

# Vector de numeración del 1 al 12
periodo = np.arange(1, 13)

# Tabla: numeración | S&P 500 | Nike Stock Price
tabla = pd.DataFrame({
    "No.": periodo,
    "S&P 500": sp500,
    "Nike Stock Price": nike
})

tabla
```

```python
import matplotlib.pyplot as plt

# Dispersión: X = S&P 500, Y = Nike Stock Price
plt.figure(figsize=(8, 5))
plt.scatter(sp500, nike, color="tab:blue", label="Datos")

# Línea de tendencia (regresión lineal)
m, b = np.polyfit(sp500, nike, 1)
plt.plot(sp500, m * sp500 + b, color="tab:red", label=f"Tendencia: y = {m:.4f}x + {b:.2f}")

plt.xlabel("S&P 500")
plt.ylabel("Nike Stock Price")
plt.title("Nike Stock Price vs S&P 500")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
