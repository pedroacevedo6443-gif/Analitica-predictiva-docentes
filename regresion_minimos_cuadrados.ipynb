```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Datos: x = S&P 500, y = Nike Stock Price
x = np.array([2912.43, 3044.31, 3100.29, 3271.12, 3500.31, 3363.00,
              3269.96, 3621.63, 3756.07, 3714.24, 3811.15, 3943.34])
y = np.array([87.18, 98.58, 98.05, 97.61, 111.89, 125.54,
              120.08, 134.70, 141.47, 133.59, 134.78, 140.45])
n = len(x)

# Sumatorias
sum_x, sum_y = x.sum(), y.sum()
sum_xy = (x * y).sum()
sum_x2 = (x ** 2).sum()

# Método de mínimos cuadrados
# Pendiente:  m = (nΣxy − ΣxΣy) / (nΣx² − (Σx)²)
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)

# Ordenada al origen:  b = (Σy − mΣx) / n   (equivale a ȳ − m·x̄)
b = (sum_y - m * sum_x) / n

print(f"Pendiente (m)        = {m:.6f}")
print(f"Ordenada al origen (b) = {b:.4f}")
print(f"Función: y = {m:.6f}·x + ({b:.4f})")
```

```python
# Residuales: e = y real − y estimada
y_est = m * x + b
residual = y - y_est

res = pd.DataFrame({
    "x": x,
    "y real": y,
    "y estimada (mx+b)": y_est,
    "residual e = y − ŷ": residual,
    "e²": residual ** 2
})
res.index = range(1, n + 1)

# Fila de sumatorias
res_tabla = pd.concat([res, pd.DataFrame([res.sum()], index=["Σ"])])
display(res_tabla.round(4))

sse = (residual ** 2).sum()
print(f"Σ residuales        = {residual.sum():.6f}  (debe ser ≈ 0)")
print(f"Suma de cuadrados de residuales (SSE) = {sse:.4f}")
print(f"Error estándar de estimación = {np.sqrt(sse / (n - 2)):.4f}")

# Verificación con numpy
m_np, b_np = np.polyfit(x, y, 1)
print(f"\nVerificación np.polyfit: m = {m_np:.6f}, b = {b_np:.4f}")
```

```python
# Gráfica: puntos, recta de mínimos cuadrados y residuales
plt.figure(figsize=(10, 6))
plt.scatter(x, y, s=120, color="tab:blue", edgecolor="white", zorder=3, label="Datos reales")

x_line = np.linspace(x.min(), x.max(), 100)
plt.plot(x_line, m * x_line + b, color="tab:red", linewidth=2.5,
         label=f"y = {m:.4f}x + ({b:.2f})")

# Segmentos verticales = residuales
plt.vlines(x, y_est, y, color="gray", linestyle="--", linewidth=1.5, label="Residuales")

plt.xlabel("S&P 500")
plt.ylabel("Nike Stock Price")
plt.title("Recta de mínimos cuadrados y residuales")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
